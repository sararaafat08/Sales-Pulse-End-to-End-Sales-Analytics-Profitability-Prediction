```python
import pandas as pd
```

```python
sales=pd.read_csv('FactSale.csv')
customer=pd.read_csv('DimCustomer.csv',skiprows=1)
city=pd.read_csv('DimCity.csv')
stock=pd.read_csv('DimStockItem.csv',skiprows=1)
date=pd.read_csv('DimDate.csv')
employee=pd.read_csv('employee.csv')
```

```python
customer.columns = customer.columns.str.strip()
customer['Credit Limit'] = customer['Credit Limit'].astype(str).str.replace(r'[^\d.]', '', regex=True)
customer['Credit Limit'] = pd.to_numeric(customer['Credit Limit'], errors='coerce').fillna(0)
customer.to_csv('Cleaned_DimCustomer.csv', index=False)

#######################################################################################################

sales['Invoice Date Key']=pd.to_datetime(sales['Invoice Date Key'])
sales=sales[sales['Quantity']>0]

#city=city.drop(columns=['Location'])
stock.columns = stock.columns.str.strip()
stock['Unit Price'] = stock['Unit Price'].astype(str).str.replace(r'[^\d.]', '', regex=True)
stock['Unit Price'] = pd.to_numeric(stock['Unit Price'], errors='coerce').fillna(0)
stock['Recommended Retail Price'] = stock['Recommended Retail Price'].astype(str).str.replace(r'[^\d.]', '', regex=True)
stock['Recommended Retail Price'] = pd.to_numeric(stock['Recommended Retail Price'], errors='coerce').fillna(0)
if 'Color' in stock.columns:
    stock = stock.drop(columns=['Color'])
stock.to_csv('Cleaned_StockItem.csv', index=False)

########################################################################################

employee.columns = employee.columns.str.strip() 
if 'Employee' in employee.columns:
    employee['Employee'] = employee['Employee'].astype(str).str.strip()
if 'Photo' in employee.columns:
    employee = employee.drop(columns=['Photo'])
if 'Is Salesperson' in employee.columns:
    employee['Is Salesperson'] = employee['Is Salesperson'].astype(bool)

employee.to_csv('Cleaned_Employee.csv', index=False)

########################################################################################

date.columns = date.columns.str.strip()
date['Date'] = pd.to_datetime(date['Date'])
date['Year'] = date['Date'].dt.year
date['Month_Name'] = date['Date'].dt.month_name()
sales['Delivery Date Key'] = pd.to_datetime(sales['Delivery Date Key'], errors='coerce')

sales.to_csv('Cleaned_FactSale.csv', index=False)
date.to_csv('Cleaned_DimDate.csv', index=False)
```

```python
customer.to_csv('Cleaned_DimCustomer.csv', index=False)
sales.to_csv('Cleaned_FactSale.csv', index=False)
stock.to_csv('Cleaned_StockItem.csv', index=False)
employee.to_csv('Cleaned_Employee.csv', index=False)
date.to_csv('Cleaned_DimDate.csv', index=False)
city.to_csv('Cleaned_DimCity.csv', index=False)
```
